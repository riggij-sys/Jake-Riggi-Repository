```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
#this cell imports our numpy, matplotlib, and scipy stats data frames so we are able to do interesting things with our data
```

```python
url="https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv"
url2='https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv'

v_rest_sst=np.loadtxt(url)
v_rest_pvalb=np.loadtxt(url2)
#this cell turns our raw csv file into numpy arrays
```

```python
v_rest_sst_mean=np.mean(v_rest_sst)
v_rest_pvalb_mean=np.mean(v_rest_pvalb)
print(v_rest_sst_mean)
print(v_rest_pvalb_mean)
ind_t_test=stats.ttest_ind(v_rest_sst, v_rest_pvalb)
print(ind_t_test)
#This cell finds the mean resting potential for our 2 neurons, and does an independent samples t test to determine the liklihood that the differnece between our data wa obtained by chance
```

```python

fig, ax = plt.subplots(1,2)
ax[0].hist(v_rest_sst)
if ind_t_test.pvalue<.05:
  ax[1].hist(v_rest_pvalb)
ax[0].set_ylabel('frequency')
ax[0].set_xlabel('resting potential')
ax[1].set_ylabel('frequency')
ax[1].set_xlabel('resting potential')
#this code creates our two subplotted histogram and lables our axes
```

```python
def vrestHist():
  """
  This function makes a histogram of the resting potentials of sst and pvalvb
  """
  fig, ax = plt.subplots(1,2)
  ax[0].hist(v_rest_sst)
  if ind_t_test.pvalue<.05:
    ax[1].hist(v_rest_pvalb)
  ax[0].set_ylabel('frequency')
  ax[0].set_xlabel('resting potential')
  ax[1].set_ylabel('frequency')
  ax[1].set_xlabel('resting potential')

#this code cell turns our histogram creating code into a function
```

```python
vrestHist()
#this code cell runs our function that creates the histogram
```

```python
v_rest_sst_len=len(v_rest_sst)
v_rest_pvalb_len=len(v_rest_pvalb)

for i in range(v_rest_sst_len):
  if v_rest_sst[i]>-60 and v_rest_pvalb[i]>-60:
    plt.scatter(v_rest_sst[i], v_rest_pvalb[i], color='red')
  elif v_rest_sst[i]>-60 and v_rest_pvalb[i]<-60:
    plt.scatter(v_rest_sst[i], v_rest_pvalb[i], color='green')
  elif v_rest_sst[i]<-60 and v_rest_pvalb[i]<-60:
    plt.scatter(v_rest_sst[i], v_rest_pvalb[i], color='blue')
  elif v_rest_sst[i]<-60 and v_rest_pvalb[i]>-60:
    plt.scatter(v_rest_sst[i], v_rest_pvalb[i], color='purple')
plt.xlabel('Sst resting potential')
plt.ylabel('Pvalb resting potential')
#This code makes our plot of resting potentials. Becasue our 2 arrays are the same length we are able to only place one length to tell our i to run 500 times. Then, we are saying if our iteration of i is above or below -60 for our two variables to color that iteration of i a certain color.
```
