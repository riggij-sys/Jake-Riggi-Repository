```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
url='https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial1.csv'
spikes_left_trial1=np.loadtxt(url, delimiter=',')
url2='https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial1.csv'
spikes_right_trial1=np.loadtxt(url2, delimiter=',')
url3='https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial2.csv'
spikes_left_trial2=np.loadtxt(url3, delimiter=',')
url4='https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial2.csv'
spikes_right_trial2=np.loadtxt(url4, delimiter=',')
```

```python
print(len(spikes_left_trial1))
print(len(spikes_right_trial1))
print(len(spikes_left_trial2))
print(len(spikes_right_trial2))

```

```python
print(spikes_left_trial1.shape)
print(spikes_right_trial1.shape)
print(spikes_left_trial2.shape)
print(spikes_right_trial2.shape)
```

```python
print(len(spikes_left_trial1)/30)
print(len(spikes_right_trial1)/30)
print(len(spikes_left_trial2)/30)
print(len(spikes_right_trial2)/30)
```

```python
fig, ax = plt.subplots(1,2)
ax[0].plot(spikes_left_trial1, np.ones_like(spikes_left_trial1), '.')
ax[0].plot(spikes_right_trial1, 1.5 * np.ones_like(spikes_right_trial1), '.')
ax[1].plot(spikes_left_trial2, np.ones_like(spikes_left_trial2), '.')
ax[1].plot(spikes_right_trial2, 1.5 * np.ones_like(spikes_right_trial2), '.')
```

```python
ISIs_spikes_left_trial1=np.diff(spikes_left_trial1)
ISIs_spikes_right_trial1=np.diff(spikes_right_trial1)
ISIs_spikes_left_trial2=np.diff(spikes_left_trial2)
ISIs_spikes_right_trial2=np.diff(spikes_right_trial2)
```

```python
print(ISIs_spikes_left_trial1.shape)
```

```python
bins=np.arange(0, 2,.01)
plt.hist(ISIs_spikes_left_trial1, bins)
plt.xlim([0, 0.15])
plt.xlabel('ISI [s]')
plt.ylabel('Counts')
```

```python
bins=np.arange(0, 2,.01)
plt.hist(ISIs_spikes_left_trial2, bins)
plt.xlim([0, 0.15])
plt.xlabel('ISI [s]')
plt.ylabel('Counts')
```

```python
bins=np.arange(0, 2,.01)
plt.hist(ISIs_spikes_right_trial1, bins)
plt.xlim([0, 0.15])
plt.xlabel('ISI [s]')
plt.ylabel('Counts')
```

```python
bins=np.arange(0, 2,.01)
plt.hist(ISIs_spikes_right_trial2, bins)
plt.xlim([0, 0.15])
plt.xlabel('ISI [s]')
plt.ylabel('Counts')
```

```python
binsize = 1
fr_bins = np.arange(0,30,binsize)
fr_spikes_left_trial1 = np.zeros(len(fr_bins)-1)
for i in range(len(fr_bins)-1):
  spk_index = (spikes_left_trial1 > fr_bins[i]) & (spikes_left_trial1 < fr_bins[i+1])
  true_values = spikes_left_trial1[spk_index]
  fr_spikes_left_trial1[i] = (len(true_values))/binsize



binsize = 1
fr_bins = np.arange(0,30,binsize)
fr_spikes_right_trial1 = np.zeros(len(fr_bins)-1)
for i in range(len(fr_bins)-1):
  spk_index = (spikes_right_trial1 > fr_bins[i]) & (spikes_right_trial1 < fr_bins[i+1])
  true_values = spikes_right_trial1[spk_index]
  fr_spikes_right_trial1[i] = (len(true_values))/binsize

binsize = 1
fr_bins = np.arange(0,30,binsize)
fr_spikes_left_trial2 = np.zeros(len(fr_bins)-1)
for i in range(len(fr_bins)-1):
  spk_index = (spikes_left_trial2 > fr_bins[i]) & (spikes_left_trial2 < fr_bins[i+1])
  true_values = spikes_left_trial2[spk_index]
  fr_spikes_left_trial2[i] = (len(true_values))/binsize



binsize = 1
fr_bins = np.arange(0,30,binsize)
fr_spikes_right_trial2 = np.zeros(len(fr_bins)-1)
for i in range(len(fr_bins)-1):
  spk_index = (spikes_right_trial2 > fr_bins[i]) & (spikes_right_trial2 < fr_bins[i+1])
  true_values = spikes_right_trial2[spk_index]
  fr_spikes_right_trial2[i] = (len(true_values))/binsize


fig, ax = plt.subplots(1,2)
ax[0].plot(fr_bins[:-1], fr_spikes_left_trial1, color='red')
ax[0].plot(fr_bins[:-1], fr_spikes_right_trial1, color='blue')
ax[1].plot(fr_bins[:-1], fr_spikes_left_trial2, color='red')
ax[1].plot(fr_bins[:-1], fr_spikes_right_trial2, color='blue')
```
